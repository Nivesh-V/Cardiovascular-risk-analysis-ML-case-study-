# Cardiovascular Risk Analysis - 10 Classification Algorithms (Combined Notebook)

**Target variable:** `cardio` (1 = cardiovascular disease present, 0 = absent)
**Dataset:** `HeartFailureDataset.csv` (~70,000 patient records)

This notebook combines two classification notebooks into one, using **one common preprocessing pipeline** for all models:

| Part | Algorithms |
|---|---|
| **Part A** (original notebook, kept first) | Logistic Regression, K-Nearest Neighbors, Gaussian Naive Bayes, Decision Tree, Support Vector Classifier (SVC) |
| **Part B** (second notebook) | Random Forest, AdaBoost, Gradient Boosting, Bagging (Decision Tree base), MLP Neural Network |

**Pipeline:** load -> inspect -> missing values / duplicates -> outlier detection and treatment (IQR capping) -> encoding -> stratified train/test split -> scaling -> PCA -> hyper-parameter tuning and training of all 10 models -> evaluation (table, ROC-AUC, confusion matrices, tree) -> inference.

> Place `HeartFailureDataset.csv` in the same folder as this notebook. Requires scikit-learn >= 1.2.

## 1. Load the dataset

```python
import warnings, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.base import clone
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Part A models (original notebook)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.svm import SVC

# Part B models (second notebook)
from sklearn.ensemble import (RandomForestClassifier, AdaBoostClassifier,
                              GradientBoostingClassifier, BaggingClassifier)
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, roc_curve, ConfusionMatrixDisplay)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
```

```python
DATA_PATH = "HeartFailureDataset.csv"
df = pd.read_csv(DATA_PATH)
df.head()
```

### 2.1 Handling duplicates / missing values and light feature engineering

```python
# Duplicates
n_before = len(df)
df = df.drop_duplicates()
print(f"Duplicates removed: {n_before - len(df)}")

# `id` is only a row identifier, it carries no medical information
df = df.drop(columns=["id"])

# Missing values: median for numeric columns, mode otherwise (nothing to fill here, kept for robustness)
for c in df.columns:
    if df[c].isnull().any():
        df[c] = df[c].fillna(df[c].median() if pd.api.types.is_numeric_dtype(df[c]) else df[c].mode()[0])
print("Remaining missing values:", int(df.isnull().sum().sum()))

# Feature engineering: age is stored in days -> convert to years; add Body-Mass-Index
df["age_years"] = (df["age"] / 365.25).round(1)
df = df.drop(columns=["age"])
df["bmi"] = df["weight"] / (df["height"] / 100) ** 2

# Quick look at physiologically impossible blood-pressure readings
print()
print("ap_hi > 250          :", (df["ap_hi"] > 250).sum())
print("ap_hi < 60           :", (df["ap_hi"] < 60).sum())
print("ap_lo > 200          :", (df["ap_lo"] > 200).sum())
print("ap_hi <= ap_lo       :", (df["ap_hi"] <= df["ap_lo"]).sum())
print("Shape now            :", df.shape)
```

## 3. Outlier detection (IQR method) and treatment

```python
cont_cols = ["age_years", "height", "weight", "ap_hi", "ap_lo", "bmi"]

def iqr_bounds(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

rows = []
for c in cont_cols:
    lo, hi = iqr_bounds(df[c])
    n_out = int(((df[c] < lo) | (df[c] > hi)).sum())
    rows.append({"feature": c, "lower_bound": round(lo, 2), "upper_bound": round(hi, 2),
                 "n_outliers": n_out, "pct_outliers": round(100 * n_out / len(df), 2)})
outlier_table = pd.DataFrame(rows)
outlier_table
```

**Boxplots BEFORE outlier treatment**

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, c in zip(axes.ravel(), cont_cols):
    sns.boxplot(y=df[c], ax=ax, color="#e07b7b")
    ax.set_title(f"{c} (before)")
plt.suptitle("Boxplots before outlier treatment", fontsize=14)
plt.tight_layout()
plt.show()
```

**Treatment:** values outside the IQR fences are **capped (winsorised)** at the fence rather than deleted. This keeps all ~70,000 observations and removes the influence of impossible values.

```python
df_clean = df.copy()
for c in cont_cols:
    lo, hi = iqr_bounds(df[c])          # fences computed on the original data
    df_clean[c] = df_clean[c].clip(lower=lo, upper=hi)

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, c in zip(axes.ravel(), cont_cols):
    sns.boxplot(y=df_clean[c], ax=ax, color="#6fb98f")
    ax.set_title(f"{c} (after)")
plt.suptitle("Boxplots after outlier treatment", fontsize=14)
plt.tight_layout()
plt.show()
```

## 4. Encoding, EDA, split, scaling and dimensionality reduction

```python
# --- Encoding ---------------------------------------------------------------
# gender: 1/2 -> 0/1 (binary encoding)
df_clean["gender"] = df_clean["gender"].map({1: 0, 2: 1})
# cholesterol and gluc are ORDINAL (1 = normal, 2 = above normal, 3 = well above normal)
# -> the existing integer order is already the correct ordinal encoding.
# any remaining text columns (none in this dataset) would be one-hot encoded
obj_cols = df_clean.select_dtypes(include=["object", "category"]).columns.tolist()
if obj_cols:
    df_clean = pd.get_dummies(df_clean, columns=obj_cols, drop_first=True)

print("Non-numeric columns left:", df_clean.select_dtypes(exclude="number").columns.tolist())
df_clean.dtypes
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
df_clean["cardio"].value_counts().sort_index().plot(kind="bar", ax=axes[0], color=["#6fb98f", "#e07b7b"])
axes[0].set_title("Target balance (0 = no CVD, 1 = CVD)")
axes[0].set_xticklabels(["0", "1"], rotation=0)
sns.heatmap(df_clean.corr(), annot=True, fmt=".2f", cmap="coolwarm", ax=axes[1], annot_kws={"size": 7})
axes[1].set_title("Correlation matrix")
plt.tight_layout()
plt.show()
```

### 4.1 Train / test split
Stratified 80/20 split so both sets keep the same class ratio. The split is done **before** scaling and PCA, and the scaler and PCA are fitted on the training data only, so no information from the test set leaks into training.

```python
X = df_clean.drop(columns="cardio")
y = df_clean["cardio"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE)
print("X_train:", X_train.shape, "| X_test:", X_test.shape)
print("Train class ratio:", y_train.value_counts(normalize=True).round(3).to_dict())
```

### 4.2 Feature scaling (StandardScaler) and dimensionality reduction (PCA)
PCA is variance based, so the features are standardised first. The number of components is chosen to retain **95 % of the variance**.

```python
scaler = StandardScaler().fit(X_train)
X_train_sc = scaler.transform(X_train)
X_test_sc = scaler.transform(X_test)

pca_full = PCA(random_state=RANDOM_STATE).fit(X_train_sc)
cum_var = np.cumsum(pca_full.explained_variance_ratio_)
n_comp = int(np.argmax(cum_var >= 0.95) + 1)

pca = PCA(n_components=n_comp, random_state=RANDOM_STATE).fit(X_train_sc)
X_train_pca = pca.transform(X_train_sc)
X_test_pca = pca.transform(X_test_sc)

plt.figure(figsize=(7, 4))
plt.plot(range(1, len(cum_var) + 1), cum_var, marker="o")
plt.axhline(0.95, color="red", ls="--", label="95% variance")
plt.axvline(n_comp, color="green", ls="--", label=f"{n_comp} components")
plt.xlabel("Number of components"); plt.ylabel("Cumulative explained variance")
plt.title("PCA - cumulative explained variance"); plt.legend()
plt.show()

print(f"Features before PCA: {X_train_sc.shape[1]}  ->  after PCA: {X_train_pca.shape[1]}")
print(f"Variance retained  : {pca.explained_variance_ratio_.sum():.3f}")
```

## 5. Model training with hyper-parameter tuning (all 10 models)
Every model - from both notebooks - is trained on the **same PCA-transformed training data** and evaluated on the same test set.

Each model is tuned with `GridSearchCV` (3-fold, scored on ROC-AUC) on a stratified sample of the training set (15,000 rows; 6,000 for the slow SVC) to keep run-time reasonable. The best parameters are then used to **refit the model on the full training set**, and the test set is only used for the final evaluation.
The hyper-parameter grids of the Part A models are the ones from the original notebook. If a model has no `predict_proba` (SVC is run without probability estimates for speed), its decision-function score is used for ROC-AUC.

```python
results = {}

def tune_and_fit(name, estimator, param_grid, group, search_size=15000):
    t0 = time.time()
    X_sub, _, y_sub, _ = train_test_split(X_train_pca, y_train, train_size=search_size,
                                          stratify=y_train, random_state=RANDOM_STATE)
    gs = GridSearchCV(estimator, param_grid, cv=3, scoring="roc_auc", n_jobs=-1)
    gs.fit(X_sub, y_sub)

    best = clone(estimator).set_params(**gs.best_params_)
    best.fit(X_train_pca, y_train)

    if hasattr(best, "predict_proba"):
        y_score = best.predict_proba(X_test_pca)[:, 1]
    else:                                   # e.g. SVC without probability=True
        y_score = best.decision_function(X_test_pca)

    results[name] = {
        "group": group,
        "model": best,
        "best_params": gs.best_params_,
        "cv_auc": gs.best_score_,
        "y_pred": best.predict(X_test_pca),
        "y_score": y_score,
        "train_acc": accuracy_score(y_train, best.predict(X_train_pca)),
        "time_s": time.time() - t0,
    }
    print(f"{name:<22} best params: {gs.best_params_}")
    print(f"{'':<22} CV ROC-AUC = {gs.best_score_:.4f} | total time = {time.time() - t0:.0f}s")
```

# PART A - Algorithms from the original classification notebook

### 5.1 Logistic Regression
**Purpose:** baseline linear classifier. Tuned parameter: `C`. Coefficients and odds ratios are interpreted in section 6.4.

```python
tune_and_fit(
    "Logistic Regression",
    LogisticRegression(max_iter=1000, random_state=RANDOM_STATE),
    {"C": [0.01, 0.1, 1, 10], "solver": ["liblinear"]},
    group="A")
```

### 5.2 K-Nearest Neighbors (KNN)
**Purpose:** distance-based classifier (the PCA-scaled features suit distance calculations). Tuned: `k`, weighting and distance metric.

```python
tune_and_fit(
    "KNN",
    KNeighborsClassifier(),
    {"n_neighbors": [3, 5, 7, 10, 15, 20, 30],
     "weights": ["uniform", "distance"],
     "metric": ["euclidean", "manhattan"]},
    group="A")
```

### 5.3 Gaussian Naive Bayes
**Purpose:** probabilistic classifier based on Bayes' theorem with a conditional-independence assumption (PCA components are uncorrelated, which suits this assumption). Tuned: `var_smoothing`.

```python
tune_and_fit(
    "Gaussian Naive Bayes",
    GaussianNB(),
    {"var_smoothing": np.logspace(-11, -7, 5)},
    group="A")
```

### 5.4 Decision Tree Classifier
**Purpose:** non-linear, rule-based classifier. Tuned: `max_depth`, `min_samples_split`, `min_samples_leaf` (to control complexity and overfitting). The tree plot is in section 6.4.

```python
tune_and_fit(
    "Decision Tree",
    DecisionTreeClassifier(random_state=RANDOM_STATE),
    {"max_depth": [2, 3, 5, 8, 10, 15, 20, None],
     "min_samples_split": [2, 5, 10],
     "min_samples_leaf": [1, 2, 5]},
    group="A")
```

### 5.5 Support Vector Classifier (SVC)
**Purpose:** margin-based classifier. Tuned: `C`, `kernel` (linear / RBF) and `gamma`. The grid search uses a 6,000-row sample because SVC training time grows steeply with the number of rows.

```python
tune_and_fit(
    "SVC",
    SVC(random_state=RANDOM_STATE, cache_size=1000),
    [{"C": [0.1, 1, 10], "kernel": ["linear"]},
     {"C": [0.1, 1, 10], "kernel": ["rbf"], "gamma": ["scale", "auto"]}],
    group="A", search_size=6000)
```

### 5.6 Random Forest (tuned: `n_estimators`, `max_depth`)

```python
tune_and_fit(
    "Random Forest",
    RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
    {"n_estimators": [100, 200, 300], "max_depth": [8, 12, 16]},
    group="B")
```

```python
tune_and_fit(
    "AdaBoost",
    AdaBoostClassifier(random_state=RANDOM_STATE),
    {"n_estimators": [50, 100, 200], "learning_rate": [0.1, 0.5, 1.0]},
    group="B")
```

### 5.8 Gradient Boosting (scikit-learn GBM)

```python
tune_and_fit(
    "Gradient Boosting",
    GradientBoostingClassifier(random_state=RANDOM_STATE),
    {"n_estimators": [100, 200], "learning_rate": [0.05, 0.1], "max_depth": [3, 4]},
    group="B")
```

## 8. Inference and conclusion

### 8.1 What the data showed
- **Data quality:** no missing values and no duplicates, but the blood-pressure columns contained clearly invalid readings (negative values, `ap_hi` up to 16,020, 1,236 rows where diastolic >= systolic). IQR capping fixed these without losing any of the 70,000 rows.
- **Target:** almost perfectly balanced (about 50 % / 50 %), so accuracy is a fair metric and no resampling was needed.
- **Dimensionality reduction:** PCA reduced 12 features to 10 components while keeping about 97.7 % of the variance. All 10 models were trained on exactly the same transformed data, so the comparison is like-for-like.
- **Main risk factors:** all three interpretation tools agree. Random Forest importance ranks systolic pressure `ap_hi` (about 45 %), diastolic pressure `ap_lo` (about 23 %), `age_years` (about 12 %) and `cholesterol` (about 9 %) at the top. The depth-3 tree splits first on `ap_hi`, then on age and cholesterol. In the Logistic Regression, one standard deviation increase in `ap_hi` multiplies the odds of cardiovascular disease by about 2.3, `age_years` by about 1.4, `cholesterol` by about 1.4 and `ap_lo` by about 1.2. Lifestyle variables (`smoke`, `alco`, `active`) and `gender` have almost no effect in this dataset.

### 8.3 Which model is better?
- **Best overall: the MLP Neural Network.** It has the highest ROC-AUC, accuracy and F1, the second-highest recall, a small train/test gap (0.7 %) and a short run time (about 20 s versus about 3 minutes for Gradient Boosting or Random Forest on the machine used).
- **Best among the tree ensembles: Gradient Boosting** (AUC 0.788), just behind the MLP. Random Forest and Bagging follow, but they overfit more (train/test gaps of 1.9 % and 2.5 %).
- **Part A models:** Logistic Regression (AUC 0.782), SVC (0.781) and KNN (0.780) are only about 0.8-1.0 AUC points behind the best model. Logistic Regression is the standout: it trains in under a second and its coefficients are directly interpretable, so it is the best choice if explainability matters. SVC has the highest precision (0.758) and the smallest overfitting gap, but its recall is low (0.643), meaning it misses more patients with the disease.
- **Weakest models:** Gaussian Naive Bayes (AUC 0.756, accuracy 0.695) suffers from its assumption that features are independent, which does not hold for correlated measurements like blood pressure, weight and BMI. The single Decision Tree (AUC 0.766) is clearly beaten by the ensembles built from many trees (about 2 AUC points higher), which shows the benefit of bagging and boosting.
- **Highest recall:** AdaBoost (0.701) and the MLP (0.698). If missing a sick patient is costly, they are the better options even though AdaBoost has the lower AUC.
- **The differences are small.** Seven of the ten models are within about 1 AUC point (0.780-0.789), and their ROC curves almost overlap. Without repeated cross-validation or a significance test, the order within that group should be read as "roughly tied", not as a decisive win.

### 8.4 Observations and caveats
- All good models plateau at about **72 % accuracy and about 0.78-0.79 AUC**. This points to a limit in the information in the data (only blood pressure, cholesterol, glucose, BMI, age and a few lifestyle flags; no ECG, lipid panel or family history), not to a weakness of a particular algorithm. Richer clinical features would help more than further tuning.
- For a **medical screening** use case, recall (not missing sick patients) matters more than precision. For any model, lowering the decision threshold below 0.5 would trade precision for higher recall.
- Models were trained on PCA components, which keeps the comparison fair but makes the trained models harder to read, which is why the tree, importances and odds ratios use the original features. Tree-based models do not need PCA and might gain a little without it.
- Hyper-parameters were tuned on subsamples (15,000 rows, and 6,000 for SVC) for speed and scored on ROC-AUC, so the ranking could shift slightly with full-data tuning.
- Part A models were originally written with a different preprocessing (removal of invalid rows, one-hot encoding, no PCA). Their numbers here are therefore not directly comparable with the numbers in the original notebook.

**Conclusion:** the **MLP Neural Network** is the recommended model (best ROC-AUC, accuracy and F1, low overfitting, fast), with **Gradient Boosting** as the best tree-ensemble alternative and **Logistic Regression** as the best simple, interpretable model. Blood pressure, age and cholesterol are the main drivers of cardiovascular risk in this dataset.
